## Part A – Theory Questions

### 1. Explain the difference between Mean and Median.

**Answer:**

* **Mean:** Sum of all values divided by the number of values.
* **Median:** The middle value when the data is arranged in ascending or descending order.
* Mean is more affected by outliers, while Median is less affected by outliers.

---

### 2. What is Variance?

**Answer:**

Variance measures how far the data values are spread from the mean.

A higher variance means the data is more spread out, while a lower variance means the data is closer to the mean.

---

### 3. What is Standard Deviation?

**Answer:**

Standard deviation measures the amount of variation or spread in a dataset.

It is the square root of variance.

---

### 4. Explain Bayes' Theorem with an example.

**Answer:**

Bayes' Theorem is used to calculate the probability of an event based on previously known information.

Example:

If a medical test is positive, Bayes' Theorem can be used to calculate the probability that the person actually has the disease.

Formula:

**P(A|B) = P(B|A) × P(A) / P(B)**

---

### 5. What is the difference between Population and Sample?

**Answer:**

* **Population:** The complete group being studied.
* **Sample:** A smaller subset selected from the population.

Example:

If we want to study the marks of all students in a college:

* All students = Population
* 100 selected students = Sample

---

### 6. Explain Correlation and Covariance.

**Answer:**

* **Correlation:** Measures the strength and direction of the relationship between two variables. Its value ranges from **-1 to +1**.
* **Covariance:** Measures the direction in which two variables change together.

Positive correlation means both variables tend to increase together.

Negative correlation means one variable tends to increase while the other decreases.

---

### 7. What is Skewness?

**Answer:**

Skewness measures the asymmetry of a dataset.

* **Positive Skew:** Tail extends toward the right.
* **Negative Skew:** Tail extends toward the left.
* **Zero Skew:** Data is approximately symmetric.

---

### 8. Explain IQR.

**Answer:**

IQR stands for **Interquartile Range**.

It measures the spread of the middle 50% of the data.

**IQR = Q3 - Q1**

Where:

* Q1 = First Quartile
* Q3 = Third Quartile

---

### 9. What is a p-value?

**Answer:**

A p-value helps determine whether there is enough statistical evidence to reject the null hypothesis.

Commonly:

* **p-value < 0.05:** Reject the null hypothesis.
* **p-value ≥ 0.05:** Fail to reject the null hypothesis.

---

### 10. What is the difference between Type I and Type II Error?

**Answer:**

* **Type I Error:** Rejecting a true null hypothesis.
* **Type II Error:** Failing to reject a false null hypothesis.

| Error         | Meaning        |
| ------------- | -------------- |
| Type I Error  | False Positive |
| Type II Error | False Negative |

### 16. Detect Outliers using the IQR Method.

**Answer:**

```python
import numpy as np

data = np.array([10, 12, 15, 18, 20, 22, 25, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Outliers:", outliers)
```

### 23. Calculate the Confidence Interval for a sample mean.

**Answer:**

```python
import numpy as np
from scipy import stats

data = [10, 12, 14, 15, 13, 16, 11, 15]

mean = np.mean(data)

confidence_interval = stats.t.interval(
    0.95,
    df=len(data) - 1,
    loc=mean,
    scale=stats.sem(data)
)

print("Mean:", mean)
print("95% Confidence Interval:", confidence_interval)
```

### 24. Perform a One-Sample Hypothesis Test.

**Question:**

Test whether the mean of the following data is significantly different from 50.

**Answer:**

```python
from scipy.stats import ttest_1samp

data = [52, 55, 53, 54, 56, 51, 50, 57]

t_stat, p_value = ttest_1samp(data, 50)

print("t-statistic:", t_stat)
print("p-value:", p_value)

if p_value < 0.05:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

### 26. Find and remove outliers using the IQR method.

**Question:**

```python
data = [10, 12, 15, 18, 20, 22, 25, 100]
```

**Answer:**

```python
import numpy as np

data = np.array([10, 12, 15, 18, 20, 22, 25, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

clean_data = data[
    (data >= lower_bound) & 
    (data <= upper_bound)
]

print("Original Data:", data)
print("Clean Data:", clean_data)
```
