## probability distribution

A `probability distribution` tells us how the possible values of a random variable are distributed and how likely each value is.

For example, if we roll a die:

X = 1, 2, 3, 4, 5, 6

Each value has probability:

$$ P(X=x)=\frac{1}{6} $$

Different situations require different probability distributions.

## Bernoulli Distribution

**Definition:**

A Bernoulli distribution represents an experiment that has exactly two possible outcomes.

Usually we call them:

Success → 1

Failure → 0

**Examples:**

Coin → Heads / Tails

Student → Pass / Fail

Product → Defective / Not defective

Email → Spam / Not spam

**Important**

A Bernoulli experiment happens only once.

**For example:**

Toss a coin one time.

Possible results:

Heads → 1

Tails → 0


**Formula**

If:

$$ P(X=1)=p $$

then:

$$ P(X=0)=1-p $$

The probability mass function is:

$$ P(X=x)=p^x(1-p)^{1-x} $$

where:

x = 0 or 1

p = probability of success

**Example**

Suppose a student has a 70% probability of passing an exam.

P(Pass) = 0.7

P(Fail) = 0.3

We can represent:

Pass → 1

Fail → 0

```python
p = 0.7

x = [0, 1]

probabilities = [
    1 - p,
    p
]

print("P(Fail):", probabilities[0])
print("P(Pass):", probabilities[1])
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

x = [0, 1]
probabilities = [0.3, 0.7]

plt.bar(x, probabilities)

plt.xticks([0, 1], ["Failure", "Success"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

plt.show()
```

### Real-world example

A manufacturing company checks whether a product is defective:

Defective     → 1

Not defective → 0

This is a Bernoulli experiment.

**AI/ML Example**

In a binary classification problem:

Spam     → 1

Not Spam → 0

A model may predict:

P(Spam) = 0.85

The underlying target variable is binary, which is commonly modeled using a Bernoulli distribution.

## Binomial Distribution

**Definition:**

A `Binomial distribution` counts the number of successes in a fixed number of independent Bernoulli trials, when each trial has the same probability of success.

**Bernoulli vs Binomial**

**This is very important:**

Bernoulli → One trial

Binomial → Multiple Bernoulli trials

**Example:**

Toss coin once → Bernoulli

Toss coin 10 times and count Heads → Binomial

Formula

$$ P(X=k)= \binom{n}{k} p^k(1-p)^{n-k} $$

Where:

n = number of trials

k = number of successes

p = probability of success

1-p = probability of failure

pmf() means **Probability Mass Function**

It gives the probability of a specific value.

```python
from scipy.stats import binom

n = 10
p = 0.5

x = np.arange(0, n + 1)

probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)

plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")

plt.show()
```

**AI/ML Example**

`Binary classification` experiments can involve repeated independent trials.

**For example:**

Out of 100 independent predictions, how many are correct?

If each prediction has the same probability of being correct, a binomial model can describe the count of correct predictions.

## Poisson Distribution

**Definition:**

The Poisson distribution models the number of times an event occurs in a fixed interval of time or space, when events occur at a known average rate and the usual Poisson assumptions are reasonable.

Examples:

Number of calls received per hour

Number of website requests per minute

Number of accidents per month

Number of customers arriving in 10 minutes

**Formula**

$$ P(X=k)= \frac{e^{-\lambda}\lambda^k}{k!} $$

Where:

λ = average number of events

k = number of events we want

e ≈ 2.718

**Example**

Suppose a call center receives an average of 4 calls per hour.

What is the probability of receiving exactly 2 calls in an hour?

λ = 4

k = 2

$$ P(X=2)= \frac{e^{-4}4^2}{2!} $$

Approximately:

0.1465

```python
from scipy.stats import poisson

lambda_value = 4
k = 2

probability = poisson.pmf(k, lambda_value)

print("Probability:", probability)

```

```python
lambda_value = 4

x = np.arange(0, 15)

probabilities = poisson.pmf(x, lambda_value)

plt.bar(x, probabilities)

plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")

plt.show()
```

### Real-world example

A website receives an average of 100 requests per second.

We could use a Poisson model to study the number of requests arriving in a fixed time interval, provided the model's assumptions are appropriate.

### AI/ML Example

Poisson distributions can be useful when modeling count data.

**For example:**

Number of customer support tickets per hour

Number of purchases per day

Number of events detected by a system

Poisson regression is a machine-learning/statistical modeling technique specifically designed for count outcomes.

## Uniform Distribution

**Definition:**

A uniform distribution is a distribution where values in a specified range have equal probability density in the continuous case.

For a continuous uniform distribution:

Minimum = a

Maximum = b

The probability density is constant between a and b.

**Formula**

For:

$$ a\leq x\leq b $$

the probability density function is:

$$ f(x)=\frac{1}{b-a} $$

**Example**

Suppose a random number is selected between:

0 and 10

Every interval of the same width has the same probability.

**For example:**

0–1

2–3

7–8

Each interval has the same probability because each has width 1.

```python
from scipy.stats import uniform

a = 0
b = 10

x = np.linspace(a, b, 1000)

probabilities = uniform.pdf(x, loc=a, scale=b-a)

plt.plot(x, probabilities)

plt.xlabel("Value")
plt.ylabel("Probability Density")
plt.title("Uniform Distribution")

plt.show()
```

`pdf()` means Probability Density Function

For continuous distributions, it describes the density rather than the probability of one exact point.

### Real-world example

Suppose a machine takes a randomly varying amount of time between 5 and 10 seconds, and every value in that interval is modeled as equally likely.

That can be represented with a uniform distribution.

### AI/ML Example

Uniform distributions are often useful for random initialization or sampling.

## Normal Distribution

**Definition:**

The normal distribution is a continuous probability distribution with a characteristic bell shape.

It is:

Symmetric

Centered around the mean

Controlled by mean and standard deviation

### Formula

The probability density function is:

$$ f(x)= \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$

Where:

μ = mean
σ = standard deviation

```python
from scipy.stats import norm

mean = 70
std = 10

x = np.linspace(30, 110, 1000)

y = norm.pdf(x, mean, std)

plt.plot(x, y)

plt.xlabel("Score")
plt.ylabel("Probability Density")
plt.title("Normal Distribution")

plt.show()
```

## Exponential Distribution

**Definition:**

The exponential distribution is commonly used to model the waiting time until the next event in a process with a constant event rate.

**For example:**

How long until the next customer arrives?

or:

How long until the next phone call?

### Formula

The probability density function is:

$$ f(x)=\lambda e^{-\lambda x} $$

for:

$$ x\geq0 $$

Where:

λ = rate of events

x = waiting time

```python
from scipy.stats import expon

lambda_value = 2

x = np.linspace(0, 3, 1000)

y = expon.pdf(
    x,
    scale=1 / lambda_value
)

plt.plot(x, y)

plt.xlabel("Waiting Time (hours)")
plt.ylabel("Probability Density")
plt.title("Exponential Distribution")

plt.show()
```

### Real-world example

Waiting time:

Customer arrives
        
      ↓

    WAIT

      ↓

Next customer arrives

The time spent waiting can be modeled using an exponential distribution when the underlying arrival process is appropriately modeled as Poisson.

### AI/ML Example

The exponential distribution can be used in:

- Survival analysis
- Reliability modeling
- Time-to-event modeling
- Modeling waiting times

To remember:

| Distribution    | Type                | Main Question                             |
| --------------- | ------------------- | ----------------------------------------- |
| **Bernoulli**   | Discrete            | Success or failure?                       |
| **Binomial**    | Discrete            | How many successes in `n` trials?         |
| **Poisson**     | Discrete            | How many events occur in an interval?     |
| **Uniform**     | Continuous/Discrete | Are outcomes equally likely?              |
| **Normal**      | Continuous          | How are values distributed around a mean? |
| **Exponential** | Continuous          | How long until the next event?            |


### Where It Is Used

- Bernoulli Distribution: Used when there are only two possible outcomes, such as success/failure, yes/no, or pass/fail.

- Binomial Distribution: Used to find the number of successes in a fixed number of independent trials, such as how many customers purchase out of 100 customers.

- Poisson Distribution: Used to count how many times an event occurs in a fixed time or space, such as calls received per hour.

- Uniform Distribution: Used when values within a given range are equally likely, such as generating random numbers within a range.

- Normal Distribution: Used to model data that tends to cluster around an average, such as heights, measurement errors, and exam scores.

- Exponential Distribution: Used to model the waiting time until the next event, such as the time until the next customer arrives.
