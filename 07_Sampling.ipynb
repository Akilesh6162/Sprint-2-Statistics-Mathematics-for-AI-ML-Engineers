## Sampling

**Definition**

Sampling is the process of selecting a smaller group of individuals (sample) from a larger group (population) to study and draw conclusions about the entire population.

`Key Terms`

- Population: The complete group we want to study.
- Sample: A smaller group selected from the population.
- Sampling: The process of selecting the sample.

**Example**

Suppose a college has 5,000 students and we want to know their average study hours.

Instead of asking all 5,000 students, we select 500 students and calculate their average study hours.

Here:

`Population` = 5,000 students

`Sample` = 500 students

`Sampling` = Process of selecting the 500 students

**Why is Sampling Used?**

- Saves time
- Reduces cost

Makes data collection easier

Useful when studying the entire population is difficult or impossible

## Random Sampling

**Definition**

Random Sampling is a sampling method where every individual in the population has an equal chance of being selected.

**Example**

A company has 1,000 employees and wants to survey 100 employees.

If 100 employees are selected randomly, every employee has an equal chance of being chosen.

```python
import random

employees = list(range(1, 1001))

sample = random.sample(employees, 100)

print(sample)
```

**Advantages**
  
- Simple to understand
- Reduces personal bias
- Every member has an equal chance of selection

**Disadvantage**
    
- Requires a complete list of the population.
- Random selection may not represent smaller groups properly.

## Stratified Sampling

**Definition**

Stratified Sampling divides the population into different groups called strata based on a specific characteristic, and then randomly selects samples from each group.

**Example**

A college has:

60% undergraduate students

40% postgraduate students

If we want a sample of 100 students, we can select:

60 undergraduate students

40 postgraduate students

This ensures both groups are represented.

```python
import pandas as pd

data = pd.DataFrame({
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Department": ["IT", "IT", "CSE", "CSE", "ECE", "ECE"]
})

sample = data.groupby("Department").sample(n=1, random_state=42)

print(sample)
```

**When is it Used?**

Stratified sampling is useful when the population contains different groups that need proper representation.

## Cluster Sampling

**Definition**

Cluster Sampling divides the population into groups called clusters, randomly selects some clusters, and studies all or many individuals within the selected clusters.

**Example**

A city has 100 schools.

Instead of selecting students from every school:

- Divide schools into clusters.
- Randomly select 10 schools.
- Survey students from those selected schools.

Here, the schools are clusters.


**Difference from Stratified Sampling**

| Stratified Sampling                 | Cluster Sampling                                        |
| ----------------------------------- | ------------------------------------------------------- |
| Select samples from **every group** | Select **some groups**                                  |
| Groups should be internally similar | Each cluster ideally represents a mix of the population |
| Focus is on representation          | Focus is often on reducing cost and effort              |


## Systematic Sampling

**Definition**

Systematic Sampling selects individuals from a population at a fixed interval.

The sampling interval is commonly calculated as:

$$ k = \frac{N}{n} $$

Where:

N = Population size

n = Sample size

k = Sampling interval

**Example**

Suppose:

Population = 1,000

Sample size = 100

$$ k = \frac{1000}{100} = 10 $$

Choose a random starting point, then select every 10th person.

For `example:`

Starting point = 5

5, 15, 25, 35, 45, 55, ...

```python
population = list(range(1, 101))

k = 10
start = 5

sample = population[start - 1::k]

print(sample)
```

**When is it Used?**

It is useful when the population is arranged in an ordered list and we want a simple and evenly distributed sample.

## Sampling Bias

**Definition**

Sampling Bias occurs when the selected sample does not properly represent the population because some members have a higher or lower chance of being selected.

**Example**

Suppose a company wants to know whether all employees are satisfied.

If the company surveys only employees from the management department, the results may not represent all employees.

This is sampling bias.

**Common Causes**
    
- Selecting only easily accessible people
- Excluding certain groups
- Voluntary participation
- Using an inappropriate sampling method

**How to Reduce Sampling Bias?**
    
- Use random selection
- Include different groups
- Use an appropriate sampling method
- Ensure the sample represents the population

## Sampling Error

**Definition**

Sampling Error is the difference between a population parameter and the corresponding value calculated from a sample.

It happens because we study a sample instead of the entire population.

**Formula**

For a population mean and sample mean:

$$ Sampling\ Error = \bar{x} - \mu $$

Where:

$$(\bar{x}) = Sample\space mean $$

$$ (\mu) = Population\space mean $$

Example

Suppose:

Population mean = 50
Sample mean = 48

Then:

$$ Sampling\ Error = 48 - 50 = -2 $$

The absolute sampling error is:

$$ |48 - 50| = 2 $$

**Important Point**

Sampling error is not necessarily caused by a mistake. It naturally occurs because a sample may differ from the complete population.

**Sampling Bias vs Sampling Error**

| Sampling Bias                                 | Sampling Error                                                  |
| --------------------------------------------- | --------------------------------------------------------------- |
| Caused by **how the sample is selected**      | Caused by studying a **sample instead of the whole population** |
| Can systematically distort results            | Usually occurs naturally                                        |
| Can be reduced through proper sampling design | Can be reduced by increasing sample size                        |
| Example: Surveying only one department        | Sample mean differs from population mean                        |


## Sample Size

**Definition**

Sample Size is the number of individuals or observations selected from a population for a study.

**Why Do We Use Sample Size?**

We use sample size to determine how many observations are needed to get reliable and representative results without studying the entire population.

**Example**

Suppose a company has 10,000 customers and wants to conduct a survey.

Instead of asking all 10,000 customers, it may select 500 customers.

Here:

Population = 10,000 customers

Sample Size = 500 customers

Where is it Used?

**Sample size is used in:**

- Surveys and questionnaires
- Market research
- Medical and scientific studies
- Quality testing
- Statistical analysis
- Machine learning and data analysis

`What Happens If Sample Size Is Too Small?`

A very small sample may:

- Give unreliable results
- Not represent the population properly
- Have greater sampling error

`What Happens If Sample Size Is Large?`

A larger sample generally:

- Gives more reliable estimates
- Reduces sampling error
- Better represents the population

However, a larger sample also requires more time, money, and resources.

**Where it is Used:**
    

- `Sampling:` Used in surveys, market research, elections, medical studies, and quality testing when studying the entire population is difficult.

- `Random Sampling:` Used when you want every member of a population to have an equal chance of being selected, such as selecting students for a survey.

- `Stratified Sampling:` Used when the population has different groups and you want each group to be represented, such as sampling employees from
different departments.

- `Cluster Sampling:` Used when the population is spread across different locations or groups, such as selecting a few schools from a city for a student survey.

- `Systematic Sampling:` Used when you have an ordered list and want to select members at regular intervals, such as checking every 10th product on a production line.

- `Sampling Bias:` Used to identify and prevent situations where the selected sample does not properly represent the population, especially in surveys and research.

- `Sampling Error:` Used to measure or understand the difference between results obtained from a sample and the actual population value.
