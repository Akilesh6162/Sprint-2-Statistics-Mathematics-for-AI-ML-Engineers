# Notebook 13 – Statistical Data Visualization

## Data Visualization

Data visualization helps us see patterns, distributions, relationships, trends, and unusual values that may be difficult to understand from numbers alone.

#### Common imports

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

## 1. Histogram

### Purpose

A Histogram shows the distribution of a numerical variable by dividing values into intervals called bins.

For example, it can show how employee ages are distributed.

### When to use

Use a histogram when you want to understand:

- Distribution of numerical data
- Center of the data
- Spread
- Skewness
- Possible outliers
- Whether data looks approximately normal

### Advantages

- Easy to understand.
- Shows the overall distribution.
- Helps identify skewness and unusual patterns.
- Useful for large datasets.

### Limitations

- Only works directly with numerical data.
- The appearance depends on the number of bins.
- Exact individual values aren't visible.

### Python Example

```python
ages = [22, 24, 25, 26, 28, 29, 30, 31, 32, 35, 38, 40, 42]

plt.hist(ages, bins=5, edgecolor="black")

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")

plt.show()
```

### Interpretation

If most bars are concentrated around 25–35, most people in the dataset are between those ages.

If the distribution has a long tail toward larger values, it may be right-skewed.

# 2. Bar Chart

## Purpose

A Bar Chart compares values between different categories.

For example, we can compare sales of different products.

## When to Use

Use a Bar Chart for:

- Product sales
- Department counts
- Categories
- Countries
- Survey responses

## Advantages

- Easy to understand
- Good for comparing categories
- Works well with categorical data

## Limitations

- Not suitable for showing continuous distributions
- Too many categories can make the chart difficult to read
- Small differences can sometimes be difficult to identify

## Example

```python
products = ["Laptop", "Phone", "Tablet"]
sales = [500, 800, 300]

plt.bar(products, sales)

plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")

plt.show()
```

## Interpretation

The Phone has the highest sales.

The Tablet has the lowest sales.

A Bar Chart makes it easy to compare values between categories.

# 3. Line Chart

## Purpose

A Line Chart shows how a value changes over an ordered sequence, especially over time.

## When to Use

Use a Line Chart for:

- Monthly sales
- Website traffic
- Revenue
- Temperature
- Stock prices
- Time-series data

## Advantages

- Excellent for showing trends
- Makes increases and decreases easy to see
- Useful for time-series data

## Limitations

- Not suitable for unrelated categories
- Too many lines can make the chart confusing
- Connecting points can imply continuity where it may not exist

## Example

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 115, 150, 170]

plt.plot(months, sales, marker="o")

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")

plt.show()
```

## Interpretation

Sales generally increased from January to May.

There was a small decrease in March compared with February.

The Line Chart makes the overall trend easy to see.

# 4. Scatter Plot

## Purpose

A Scatter Plot shows the relationship between two numerical variables.

For example:

Experience → Salary

## When to Use

Use a Scatter Plot to identify:

- Relationships between variables
- Positive correlation
- Negative correlation
- Weak or no relationship
- Clusters
- Potential outliers

## Advantages

- Clearly shows relationships
- Helps identify outliers
- Useful for correlation analysis
- Useful before regression modeling

## Limitations

- Points can overlap in large datasets
- Usually shows only two variables at a time
- Correlation does not prove causation

## Example

```python
experience = [1, 2, 3, 5, 7, 10]
salary = [30000, 35000, 40000, 50000, 65000, 80000]

plt.scatter(experience, salary)

plt.xlabel("Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")

plt.show()
```

## Interpretation

The points generally move upward from left to right.

This indicates a positive relationship between Experience and Salary.

As experience increases, salary also tends to increase in this example.

However, this does not prove that experience causes salary to increase.

# 5. Pie Chart

## Purpose

A Pie Chart shows how different categories contribute to a whole.

It is useful for showing percentages or proportions.

## When to Use

Use a Pie Chart when:

- There are only a few categories
- You want to show percentages
- Categories represent parts of one total

## Advantages

- Easy to understand for a small number of categories
- Shows proportions visually
- Useful for simple percentage comparisons

## Limitations

- Difficult to compare similar-sized slices
- Becomes confusing with many categories
- Bar Charts are often better for precise comparisons

## Example


```python
products = ["Laptop", "Phone", "Tablet"]
sales = [500, 800, 300]

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution")

plt.show()
```

## Interpretation

The Phone represents the largest portion of total sales.

The Tablet represents the smallest portion.

The percentages show each product's contribution to the total sales.

# 6. Density Plot

## Purpose

A Density Plot shows a smooth estimate of the distribution of numerical data.

It is similar to a Histogram, but instead of bars, it uses a smooth curve.

## When to Use

Use a Density Plot to understand:

- Distribution shape
- Peaks
- Skewness
- Differences between distributions

## Advantages

- Provides a smooth representation of a distribution
- Useful for comparing multiple distributions
- Less dependent on histogram bin selection

## Limitations

- Can be harder for beginners to understand
- The smoothing method affects the shape
- Can be misleading with very small datasets

## Example

```python
ages = [22, 24, 25, 26, 28, 29, 30, 31, 32, 35, 38, 40, 42]

sns.kdeplot(ages)

plt.xlabel("Age")
plt.ylabel("Density")
plt.title("Age Density Plot")

plt.show()
```

## Interpretation

The highest part of the curve represents where observations are most concentrated.

The shape of the curve helps us understand the distribution of the data.

# 7. Box Plot

## Purpose

A Box Plot summarizes the distribution of numerical data.

It shows:

- Minimum
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum
- Potential outliers

## When to Use

Use a Box Plot to:

- Find potential outliers
- Understand data spread
- Compare distributions
- Compare multiple groups

## Advantages

- Compact summary of data
- Good for detecting potential outliers
- Easy to compare groups

## Limitations

- Does not show the complete distribution shape
- Can hide important patterns
- Points identified as outliers may not necessarily be errors

## Example

```python
salary = [30000, 32000, 35000, 36000, 38000, 40000, 42000, 100000]

plt.boxplot(salary)

plt.ylabel("Salary")
plt.title("Salary Distribution")

plt.show()
```

## Interpretation

The line inside the box represents the median.

The box represents the middle 50% of the data.

A point far away from the main box and whiskers may be a potential outlier.

In this example, 100000 may be identified as a potential outlier.

# 8. Violin Plot

## Purpose

A Violin Plot combines information from a Box Plot and a Density Plot.

It shows the distribution and density of numerical data for one or more groups.

## When to Use

Use a Violin Plot when you want to:

- Compare distributions between groups
- See the shape of distributions
- Identify areas where observations are concentrated

## Advantages

- Shows more distribution information than a basic Box Plot
- Useful for comparing groups
- Shows density and distribution shape

## Limitations

- More difficult to understand than a Box Plot
- Can be misleading with very small datasets
- Density estimation depends on smoothing

## Example

```python
data = pd.DataFrame({
    "Department": ["IT"] * 5 + ["HR"] * 5,
    "Salary": [
        50000, 55000, 60000, 62000, 65000,
        40000, 42000, 45000, 47000, 50000
    ]
})

sns.violinplot(
    x="Department",
    y="Salary",
    data=data
)

plt.title("Salary Distribution by Department")

plt.show()
```

## Interpretation

The width of the violin represents the estimated density of observations.

A wider section means more observations are concentrated around that salary range.

The plot allows us to compare the salary distributions of IT and HR.

# 9. Pair Plot

## Purpose

A Pair Plot shows relationships between multiple numerical variables in a single visualization.

It creates:

- Scatter Plots between variables
- Distribution plots along the diagonal

## When to Use

Use a Pair Plot during:

- Exploratory Data Analysis (EDA)
- Correlation analysis
- Feature relationship analysis
- Machine-learning data exploration

## Advantages

- Shows many relationships at once
- Helps identify correlations
- Can reveal clusters
- Useful for initial data exploration

## Limitations

- Becomes large when there are many features
- Can be difficult to interpret with many variables
- Mostly useful for a small number of variables

## Example;

```python
df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Experience": [1, 3, 7, 10, 15],
    "Salary": [30000, 40000, 60000, 80000, 100000]
})

sns.pairplot(df)

plt.show()
```

## Interpretation

The Pair Plot allows us to examine relationships such as:

- Age vs Experience
- Age vs Salary
- Experience vs Salary

If points form an upward pattern, the variables may have a positive relationship.

If points form a downward pattern, they may have a negative relationship.

If points appear randomly scattered, there may be little or no linear relationship.

# 10. Heatmap

## Purpose

A Heatmap represents values using different levels of color intensity.

One of the most common uses of a Heatmap in statistics is visualizing a Correlation Matrix.

## When to Use

Use a Heatmap for:

- Correlation matrices
- Feature relationships
- Frequency tables
- Missing-value patterns
- Comparing values across two dimensions

## Advantages

- Makes large tables easier to understand
- Quickly highlights high and low values
- Excellent for correlation analysis

## Limitations

- Exact values can be difficult to compare without annotations
- Color interpretation can sometimes be misleading
- Too many variables can make the chart crowded

## Example

```python
df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Experience": [1, 3, 7, 10, 15],
    "Salary": [30000, 40000, 60000, 80000, 100000]
})

correlation = df.corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

## Interpretation

Correlation values range from -1 to +1.

- +1 → Strong positive relationship
- 0 → No linear relationship
- -1 → Strong negative relationship

For example, if Experience and Salary have a correlation of 0.98, they have a very strong positive linear relationship in this dataset.

Remember:

Correlation does not mean causation.

# Final Comparison

| Visualization | Main Purpose | Best For |
|---|---|---|
| Histogram | Show distribution | Numerical data |
| Bar Chart | Compare categories | Categorical data |
| Line Chart | Show trends | Time-series data |
| Scatter Plot | Show relationships | Two numerical variables |
| Pie Chart | Show parts of a whole | Small number of categories |
| Density Plot | Show smooth distribution | Numerical data |
| Box Plot | Show spread and outliers | Numerical data |
| Violin Plot | Show distribution and density | Comparing groups |
| Pair Plot | Show multiple relationships | Exploratory Data Analysis |
| Heatmap | Show values/relationships using color | Correlation matrices |

# Which Visualization Should I Use?

### If you want to know:

**"How is my numerical data distributed?"**

→ Histogram / Density Plot

**"Are there outliers?"**

→ Box Plot

**"How do two numerical variables relate?"**

→ Scatter Plot

**"How does a value change over time?"**

→ Line Chart

**"Which category has a higher value?"**

→ Bar Chart

**"What percentage does each category represent?"**

→ Pie Chart

**"How do distributions differ between groups?"**

→ Violin Plot

**"How are many variables related?"**

→ Pair Plot / Heatmap

# Key Takeaways

- Histogram → Distribution
- Bar Chart → Category comparison
- Line Chart → Trend
- Scatter Plot → Relationship
- Pie Chart → Proportion
- Density Plot → Smooth distribution
- Box Plot → Spread and outliers
- Violin Plot → Distribution + density
- Pair Plot → Multiple relationships
- Heatmap → Correlation/values using color

The goal of visualization is not just to make charts.

The goal is to make the data easier to understand and help identify useful patterns.
