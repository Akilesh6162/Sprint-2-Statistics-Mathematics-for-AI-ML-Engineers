# Data Distribution

## What is Data Distribution?

A data distribution describes how the values in a dataset are spread or arranged.

For example:

data = [10, 20, 20, 30, 30, 30, 40, 50]

We can ask:

- Where are most values located?
- Are values evenly distributed?
- Is the data symmetric?
- Does the data have a long left or right tail?
- Are extreme values common?

We can understand these patterns using graphs and statistical measures.

## Normal Distribution

A normal distribution is a distribution where most values are concentrated around the center, and fewer values occur as we move away from the center.

A normal distribution is:

- Bell-shaped
- Symmetric
- Centered around the mean
- Mean = Median = Mode

**Example**

Suppose the heights of students are approximately:

150, 155, 160, 165, 168, 170, 171, 172, 175, 180, 185

Most values are around the middle.

We can visualize a normal distribution using np.random.normal().

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=170, scale=10, size=1000)

plt.hist(data, bins=30,edgecolor = "black")
plt.xlabel("Height")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

| Parameter   | Meaning            |
| ----------- | ------------------ |
| `loc=170`   | Mean               |
| `scale=10`  | Standard deviation |
| `size=1000` | Number of values   |


## Uniform Distribution

In a uniform distribution, values have approximately equal likelihood across a specified range.

```python
data = np.random.uniform(low=1, high=10, size=1000)

plt.hist(data, bins=20, edgecolor= "black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

## Gaussian Distribution

Gaussian distribution and normal distribution are the same thing.

The Gaussian distribution is described mainly by:
- Mean (μ) → center
- Standard deviation (σ) → spread

```python
mean = 50
std = 10

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30, edgecolor = "black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Gaussian Distribution")
plt.show()
```

*Mean controls the center, while standard deviation controls the spread.*

## Symmetric Distribution

A distribution is symmetric when the left and right sides have approximately the same shape.

```python
data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30, edgecolor = "black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Symmetric Distribution")
plt.show()
```

## Left Skew

A left-skewed distribution has a long tail extending toward the left.

It is also called negative skewness.

**Typical relationship**

Mean<Median<Mode

```python
data = 100 - np.random.exponential(scale=10, size=1000)

plt.hist(data, bins=30, edgecolor = "black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Left-Skewed Distribution")
plt.show()
```

## Right Skew

A right-skewed distribution has a long tail extending toward the right.

It is also called positive skewness.

**Typical relationship**

Mode<Median<Mean

```python
data = np.random.exponential(scale=10, size=1000)

plt.hist(data, bins=30, edgecolor = "black")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Right-Skewed Distribution")
plt.show()
```

To remember:
| Distribution | Tail                   |
| ------------ | ---------------------- |
| Left Skew    | Tail points left       |
| Symmetric    | Both sides are similar |
| Right Skew   | Tail points right      |


## Kurtosis

`Kurtosis` describes the tail behavior of a distribution, particularly how much extreme values contribute to the distribution.

In Python, SciPy can calculate kurtosis.

```python
from scipy import stats 
data = np.random.normal(50, 10, 1000)

kurtosis_value = stats.kurtosis(data)

print("Kurtosis:", kurtosis_value)
```

Three commonly discussed types

`1. Mesokurtic`

Similar tail behavior to the normal distribution.

Excess kurtosis ≈ 0

`2. Leptokurtic`

Has heavier tails and more extreme observations.

Excess kurtosis > 0

`3. Platykurtic`

Has lighter tails and fewer extreme observations.

Excess kurtosis < 0

## Empirical Rule

The Empirical Rule applies to approximately normal distributions.

It is also called the:

68–95–99.7 Rule

It tells us approximately how much data falls within 1, 2, and 3 standard deviations from the mean.

- About 68% of observations lie within 1 standard deviation of the mean.
- About 95% lie within 2 standard deviations.
- About 99.7% lie within 3 standard deviations.

```python
import numpy as np

data = np.random.normal(70, 10, 1000)

mean = np.mean(data)
std = np.std(data)

print("Mean:", mean)
print("1 SD:", mean - std, "to", mean + std)
print("2 SD:", mean - 2*std, "to", mean + 2*std)
print("3 SD:", mean - 3*std, "to", mean + 3*std)
```

Most observations in a normal distribution fall close to the mean, while values far from the mean are uncommon.
