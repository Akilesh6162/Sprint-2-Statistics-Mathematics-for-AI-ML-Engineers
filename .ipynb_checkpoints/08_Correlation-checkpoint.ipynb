# Notebook 8 – Correlation & Covariance

## Correlation

**Definition**

`Correlation` measures the strength and direction of the relationship between two variables.

It tells us whether two variables tend to change together.

**Why Do We Use It?**

We use correlation to understand whether changes in one variable are associated with changes in another variable.

**Example**

If study hours increase and exam marks also increase, they may have a positive correlation.

**Correlation Range**

Correlation coefficient is between:

$$ -1 \leq r \leq 1 $$
- +1 → Perfect positive correlation
- 0 → No linear correlation
- -1 → Perfect negative correlation

## Covariance

**Definition**

Covariance measures the direction in which two variables change together.

**Why Do We Use It?**

We use covariance to determine whether two variables tend to increase together or whether one tends to increase when the other decreases.

**Formula**
    
$$ Cov(X,Y)=\frac{\sum(X_i-\bar X)(Y_i-\bar Y)}{n-1} $$

**Interpretation**

Positive covariance → Variables tend to increase together.

Negative covariance → One tends to increase while the other decreases.

Covariance near 0 → Little evidence of a linear relationship.

```python
import numpy as np

x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

covariance = np.cov(x, y)[0, 1]

print(covariance)
```

**Important Difference**

Covariance tells us the direction, while correlation tells us both direction and standardized strength.

## Positive Correlation

**Definition**

Positive Correlation occurs when two variables tend to increase or decrease together.

**Example**

As study hours increase, exam marks may increase.

Study Hours ↑  →  Marks ↑

**Correlation**
    
$$ 0 < r \leq 1 $$
    
**Why Do We Use It?**

It is used to identify variables that move in the same direction.

## Negative Correlation

**Definition**

`Negative Correlation` occurs when one variable increases while the other tends to decrease.

**Example**

As product price increases, demand may decrease.

Price ↑  →  Demand ↓

**Correlation**

$$ -1 \leq r < 0 $$

**Why Do We Use It?**

It is used to identify variables that move in opposite directions.

## No Correlation

**Definition**

`No Correlation` means there is no meaningful linear relationship between two variables.

**Example**

A person's shoe size and their exam score generally have no meaningful linear relationship.

**Correlation**

$$ r \approx 0 $$

**Important**

A correlation of 0 means no linear correlation; it does not necessarily mean there is no possible relationship of any kind.

## Pearson Correlation

**Definition**

Pearson Correlation measures the strength and direction of a linear relationship between two numerical variables.

It is represented by r.

**Formula**
    
$$ r = \frac{Cov(X,Y)}{\sigma_X\sigma_Y} $$

Where:

Cov(X,Y) = Covariance

σX = Standard deviation of X

σY = Standard deviation of Y

```python
import pandas as pd

x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

correlation = pd.Series(x).corr(pd.Series(y))

print(correlation)
```

## Spearman Correlation

**Definition**

Spearman Correlation measures the strength and direction of the relationship between two variables based on their ranks.

It is particularly useful when the relationship is monotonic but not necessarily linear, or when the data contains outliers or ordinal/ranked data.

**Example**

Suppose students are ranked according to:

Study time

Exam performance

Spearman correlation can determine whether students with higher study-time ranks tend to have higher performance ranks.

```python
from scipy.stats import spearmanr

x = [10, 20, 30, 40, 50]
y = [15, 25, 35, 45, 55]

correlation, p_value = spearmanr(x, y)

print(correlation)
```

**When is it Used?**

Spearman correlation is used for ranked/ordinal data or when the relationship is monotonic but not necessarily linear.

### Pearson vs Spearman Correlation

| Pearson                      | Spearman                                                             |
| ---------------------------- | -------------------------------------------------------------------- |
| Measures linear relationship | Measures monotonic relationship                                      |
| Uses actual values           | Uses ranks                                                           |
| Suitable for numerical data  | Useful for ordinal/ranked data                                       |
| More sensitive to outliers   | Generally less sensitive to outliers                                 |
| Measures linear association  | Measures whether variables generally move in the same/opposite order |


## Correlation Matrix

**Definition**

A Correlation Matrix is a table that shows the correlation coefficients between multiple variables.

**Example**

Suppose we have:

- Age
- Salary
- Experience
- Spending

A correlation matrix shows the relationship between every pair.

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 25, 30, 35, 40],
    "Salary": [25000, 35000, 45000, 55000, 65000],
    "Experience": [1, 3, 5, 8, 12],
    "Spending": [2000, 3000, 4000, 4500, 5000]
})

correlation_matrix = data.corr()

print(correlation_matrix)
```

**Why Do We Use It?**

We use a correlation matrix to quickly examine relationships among many variables at the same time.

## Heatmap

**Definition**

A `Heatmap` is a visual representation of data using different shades of color to make patterns easier to identify.

For correlation analysis, a heatmap can display the correlation matrix visually.



```python
!pip install seaborn
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

correlation_matrix = data.corr()

sns.heatmap(correlation_matrix, annot=True)

plt.show()
```

**Why Do We Use It?**

A correlation heatmap makes it easier to visually identify strong positive, strong negative, and weak relationships between variables.


#### Correlation vs Covariance

| Correlation                            | Covariance                                 |
| -------------------------------------- | ------------------------------------------ |
| Measures direction and strength        | Mainly indicates direction                 |
| Range is `-1` to `+1`                  | Has no fixed range                         |
| Standardized                           | Not standardized                           |
| Easier to interpret                    | Depends on the units of variables          |
| Commonly used to compare relationships | Useful for understanding joint variability |
