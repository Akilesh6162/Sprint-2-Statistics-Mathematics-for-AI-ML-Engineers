# Notebook 1 – Introduction to Statistics

## What is Statistics?

`Statistics` is the process of collecting, organizing, analyzing, interpreting, and presenting data so that we can understand patterns and make decisions.

Statistics helps us turn data into meaningful information

For example, suppose a company has the following sales:

₹10,000

₹15,000

₹12,000

₹18,000

₹20,000

Looking at these numbers individually doesn't tell us much.

Statistics allows us to calculate things such as:

- Average sales
- Minimum sales
- Maximum sales
- Variation in sales
- Trends
- Probability of future sales




## Why Statistics is Important in AI/ML

Machine Learning works with data.

Statistics helps us understand that data before, during, and after building a model.

For example, before training a model we may want to know:

- What is the average age?
- Are there missing values?
- Are some values unusually large?
- Which features are related?
- How is the data distributed?

Statistics helps answer these questions.

Our dataset contains:

| Area | Bedrooms | Price |
| ---: | -------: | ----: |
| 1000 |        2 |  ₹40L |
| 1200 |        2 |  ₹48L |
| 1500 |        3 |  ₹65L |
| 2000 |        4 |  ₹90L |

 Statistics can help us discover that larger houses generally have higher prices.

 AI/ML Usage

Statistics is used during:

Data → Statistics → Understanding → Preprocessing → ML Model → Evaluation
 

## Statistics vs Machine Learning

Statistics and Machine Learning are closely related, but their primary goals are different.

*`Statistics`*

Statistics generally focuses on:

Understanding data and relationships between variables.

*`Machine Learning`*

Machine Learning generally focuses on:

Learning patterns from data to make predictions or decisions
    
For example:

Statistics might ask:

"Does house size have a significant relationship with price?"

Machine Learning might ask:

"Can we predict the price of a new house?"

## Descriptive Statistics

Descriptive statistics is used to summarize and describe the data we already have.


```python
import pandas as pd
```

```python
marks = pd.Series([50, 60, 70, 80, 90])

print(marks.describe())
```

For this data:

Mean = 70
Median = 70
Minimum = 50
Maximum = 90

This is called descriptive statistics because we are simply describing the data.

AI/ML usage

Descriptive statistics is one of the first things we do during Exploratory Data Analysis (EDA).

## Inferential Statistics

`Inferential statistics` is different from descriptive statistics.

Instead of simply describing the data we have, we use a sample to make conclusions about a larger population.

For example, imagine a company has 1 million customers.

It would be difficult to ask every customer whether they like a new product.

Instead, the company might select 10,000 customers and ask them.

1,000,000 customers --> Select 10,000 customers --> Collect information--> Analyze the sample-->Make an estimate about all customers

The 10,000 customers are the sample, while the 1 million customers are the population.

**AI/ML usage**

`Inferential statistics` is commonly used in:

- A/B testing
- Hypothesis testing
- Confidence intervals
- Experiments
- Model evaluation

## Population

A `population` means the complete group that we are interested in studying.

For example, suppose a company has 5,000 employees and we want to study their salaries.

Then:

Population = All 5,000 employees

The population doesn't necessarily mean "people". It can be anything.

## Sample

A `sample` is a smaller group selected from a population.

Suppose a company has 100,000 customers.

Instead of studying all 100,000:

Population = 100,000 customers

Sample = 1,000 customers

*Sample = a smaller part of the population.*

## Census

A `census` means collecting information from every member of the population.

For example, a school has 500 students.

If the school measures the height of all 500 students:

Population = 500 students

Census = Height of all 500 students

If the school measures only 50 students, that is sampling, not a census.

## Sampling

`Sampling` is the process of selecting a smaller group from a larger population.

Imagine Amazon wants to know whether customers are satisfied with a new service.

There may be millions of customers.

Instead of contacting everyone, Amazon could select a smaller group.

Population

1,000,000 customers --> Sampling --> 10,000 customers --> Survey

Sampling is extremely useful when the population is too large to study completely.

**AI/ML usage**

Sampling is often used when datasets contain millions or billions of records.

## Variables

A variable is a characteristic that can have different values.

employee information:

| Name  | Age | Salary |
| ----- | --: | -----: |
| Ravi  |  25 | 40,000 |
| Priya |  30 | 55,000 |
| Arun  |  35 | 70,000 |


*Name, Age, Salary are variables.*

**AI/ML usage**

Variables are the basic building blocks of a dataset. Some variables become features, while another variable may become the target.

## Features

A `feature` is an input variable that an ML model uses to make a prediction.

Suppose we want to predict house prices.

We could have:

- Area
- Bedrooms
- Age of House
- Location

These can be features.

The model uses them to predict:

- Price



```python
import pandas as pd

df = pd.DataFrame({
    "Area": [1000, 1500, 2000],
    "Bedrooms": [2, 3, 4],
    "Age": [10, 5, 2],
    "Price": [40, 65, 90]
})

X = df[["Area", "Bedrooms", "Age"]]
y = df["Price"]

print("Features:")
print(X)

print("\nTarget:")
print(y)
```

- X → Features
- y → Target

**AI/ML usage**

Features are the inputs given to an ML model.

## Observation

An observation is one individual record in a dataset.

For example:
| Age | Salary |
| --: | -----: |
|  25 | 40,000 |
|  30 | 50,000 |
|  35 | 65,000 |

Each row is an observation.

## Dataset

A `dataset` is a collection of observations and variables.

| Age | Experience | Salary |
| --: | ---------: | -----: |
|  22 |          0 | 25,000 |
|  25 |          2 | 40,000 |
|  30 |          5 | 60,000 |
|  35 |          8 | 75,000 |


- Rows    → Observations
- Columns → Variables
- Entire table → Dataset



```python
import pandas as pd

dataset = pd.DataFrame({
    "Name" : ["Akilesh", "Maheswari", "Elumalai"],
    "Age" : [23, 45, 48],
    "City" : ["Tiruvannamalai", "Chennai", "Bangalore"]
})

print(dataset)
```

In Machine Learning, datasets are used for:

- Training
- Validation
- Testing

## Quantitative Data

`Quantitative data` is data that represents numerical measurements or counts.

Examples:

Age = 25

Salary = 50,000

Height = 175.5 cm

Weight = 70 kg

Number of orders = 20

Because these values have numerical meaning, we can perform mathematical operations on them when appropriate.

AI/ML usage

Examples of quantitative ML features:

- Age
- Income
- Temperature
- Height
- Weight
- Number of purchases

```python
import numpy as np

salary = np.array([3000,20000,34324,2323423,234342558])

print("Mean:",np.mean(salary))
```

## Qualitative Data

`Qualitative data` represents categories, labels, or qualities rather than numerical measurements.

Example :

Payment Method

- UPI
- Cash
- Card
- UPI
- Card

We cannot calculate an average of these categories.

```python
import pandas as pd 

payment_method = pd.Series([ "UPI",
    "Cash",
    "Card",
    "UPI",
    "Card"])

print(payment_method.value_counts())
```

The output tells us how many times each category appears.

**AI/ML usage**

ML algorithms often need categorical data to be converted into numerical representations.

Common methods include:

- One-hot encoding
- Label encoding
- Target encoding

## Continuous Data

`Continuous data` can take many possible values within a range.

Examples:

Height = 175.5 cm

Weight = 68.72 kg

Temperature = 36.7°C

Time = 10.52 seconds

For example, a person's weight could be:

68 kg

68.1 kg

68.15 kg

68.157 kg

68.1578 kg
...

There can be increasingly precise values.

AI/ML usage

Continuous data is commonly found in:

- Sensor measurements
- Temperature data
- Financial measurements
- Medical measurements
- Scientific data

## Discrete Data

Discrete data consists of values that can be counted individually.

Examples:

Number of students = 50

Number of cars = 3

Number of purchases = 10

Number of website visits = 25

For example, a shop can have:

1 customer

2 customers

3 customers

But we don't normally say:

2.5 customers

Continuous → Measured

Discrete   → Counted

**AI/ML usage**

Examples include:

Number of purchases

Number of complaints

Number of website visits

Number of children

## Structured Data

Structured data is data that follows a predefined organization, usually in the form of rows and columns.

| ID | Name  | Age | Salary |
| -: | ----- | --: | -----: |
|  1 | Ravi  |  25 | 40,000 |
|  2 | Priya |  30 | 55,000 |
|  3 | Arun  |  28 | 48,000 |

The structure is clearly defined.

```python
import pandas as pd

df = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [25, 30, 28],
    "Salary": [40000, 55000, 48000]
})

print(df)
```

**AI/ML usage**

Structured data is widely used in:

Fraud detection

Customer prediction

Sales prediction

Credit scoring

Recommendation systems
