## Statistical tests

`Statistical tests` help us decide whether an observed difference or relationship in data is statistically significant or could have happened due to random variation.

A common rule is:

- p-value < 0.05 → Reject the Null Hypothesis
- p-value ≥ 0.05 → Fail to reject the Null Hypothesis

The 0.05 threshold is a common choice, not a universal rul

## T-Test

### Definition

A T-Test is used to determine whether there is a statistically significant difference between the means of one or two groups.

**There are three common types:**

- One-sample T-Test – compare a sample mean with a known value.
- Independent T-Test – compare means of two independent groups.
- Paired T-Test – compare two measurements from the same subjects.

### Assumptions

- Data is approximately normally distributed.
- Observations are independent for an independent t-test.
- For the standard independent t-test, the two groups should have similar variances.
- Alternatively, Welch's t-test can be used when variances are unequal.
- Data should generally be continuous.

### When to use

**Use a T-Test when:**

- You are comparing means.
- The sample size is relatively small.
- Population standard deviation is unknown.

### When not to use

**Don't use it when:**

- Data is strongly non-normal and sample size is small.
- You are comparing more than two independent groups → use ANOVA.
- Your data is categorical → consider a Chi-Square Test.
- A non-parametric alternative is more appropriate → consider Mann-Whitney U.

### Business Example

A company wants to know whether a new training program changed employee productivity.

- Before training: productivity scores
- After training: productivity scores

A paired T-Test can determine whether the average productivity changed significantly.

### Python Example

```python
from scipy.stats import ttest_rel

before = [70, 72, 68, 75, 71, 69, 73]
after  = [75, 78, 72, 79, 76, 74, 77]

t_stat, p_value = ttest_rel(before, after)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

### Interpretation 

```python
if p_value < 0.05:
    print("Significant difference")
else:
    print("No significant difference")
```

If p_value < 0.05, there is evidence that the average productivity before and after training differs.

## Z-Test

### Definition

A `Z-Test` determines whether a sample mean is significantly different from a population mean, or whether two means/proportions differ significantly, under conditions where a normal/Z approximation is appropriate.

**A common one-sample formula is:**

$$ Z = \frac{\bar{x}-\mu}{\sigma/\sqrt{n}} $$

**Where:**

x̄ = sample mean

μ = population mean

σ = population standard deviation

n = sample size

### Assumptions

- Observations are independent.
- Data is approximately normally distributed, or the sample is large enough for the normal approximation to be reasonable.
- For the classic one-sample Z-test, the population standard deviation is known.

### When to use

**Use a Z-Test when:**

- Population standard deviation is known.
- Sample size is large.
- You are testing means or proportions using a normal approximation.

### When not to use

**Don't use it when:**

- Population standard deviation is unknown and the sample is small → usually use a T-Test.
- Data is strongly non-normal with a small sample.
- Your outcome is categorical and the assumptions for a Z-test aren't satisfied.

### Business Example

A manufacturer claims that its batteries have an average lifetime of 100 hours.

You take a large sample of batteries and want to determine whether their average lifetime differs from 100 hours.

### Python Example

```python
from statsmodels.stats.weightstats import ztest

battery_life = [101, 99, 102, 98, 100, 103, 101, 99, 102, 100]

z_stat, p_value = ztest(
    battery_life,
    value=100
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

### Interpretation

```python
if p_value < 0.05:
    print("Significant difference")
else:
    print("No significant difference")
```

A small p-value indicates evidence that the sample mean differs from the hypothesized mean of 100 hours.

## Chi-Square Test

### Definition

A Chi-Square Test is used mainly with categorical data to determine whether observed frequencies differ from expected frequencies or whether two categorical variables are associated.

**Two common forms are:**

- Chi-Square Goodness-of-Fit Test
- Chi-Square Test of Independence

### Assumptions

- Data consists of frequency/counts.
- Observations are independent.
- Categories are mutually exclusive.
- Expected frequencies should generally be sufficiently large.

### When to use

**Use it when:**

- Working with categorical variables.
- Comparing observed and expected counts.
- Testing whether two categorical variables are related.

### When not to use

**Don't use it when:**

- Your data is continuous and you are comparing means.
- Expected frequencies are too small for the chi-square approximation to be reliable.
- Observations are not independent.

### Business Example

An e-commerce company wants to know whether payment method is related to customer age group.

| Age Group | UPI | Card | Cash |
| --------- | --: | ---: | ---: |
| Young     | 120 |   50 |   30 |
| Middle    |  80 |   90 |   20 |
| Older     |  40 |   70 |   50 |

A Chi-Square Test of Independence can test whether age group and payment method are associated.

### Python Example



```python
from scipy.stats import chi2_contingency

data = [
    [120, 50, 30],
    [80, 90, 20],
    [40, 70, 50]
]

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-square:", chi2)
print("P-value:", p_value)
```

### Interpretation

```python
if p_value < 0.05:
    print("Variables are associated")
else:
    print("No significant association")
```

If p_value < 0.05, there is evidence of an association between age group and payment method.

## ANOVA

### Definition

ANOVA (Analysis of Variance) determines whether the means of three or more groups are significantly different.

Instead of performing many T-Tests, ANOVA tests the groups together.

The basic idea is:

$$ F = \frac{\text{Between-group variation}} {\text{Within-group variation}} $$


### Assumptions

- Observations are independent.
- Each group's data is approximately normally distributed.
- Groups have reasonably similar variances.
- The dependent variable is continuous.
  
### When to use

**Use ANOVA when:**

- Comparing means across 3 or more groups.
- Groups are independent.
- Data approximately satisfies ANOVA assumptions.

### When not to use

**Don't use it when:**

You only have two groups → a T-Test may be appropriate.
Data is strongly non-normal and assumptions are not reasonable → consider a non-parametric alternative.
The outcome is categorical.
Business Example

A company tests three different advertising campaigns:

Campaign A

Campaign B

Campaign C

It wants to know whether the average sales generated differs between the campaigns.

### Python Example

```python
from scipy.stats import f_oneway

campaign_A = [100, 105, 110, 108, 102]
campaign_B = [120, 125, 130, 128, 122]
campaign_C = [140, 145, 150, 148, 142]

f_stat, p_value = f_oneway(
    campaign_A,
    campaign_B,
    campaign_C
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
```

### Interpretation

```python
if p_value < 0.05:
    print("At least one group mean is different")
else:
    print("No significant difference")
```

**Important:** ANOVA tells you that at least one group differs. It does not tell you exactly which groups differ. A post-hoc test, such as Tukey's HSD, can be used afterward.

## Mann-Whitney U Test

### Definition

The Mann-Whitney U Test is a non-parametric test used to compare two independent groups.

It is commonly used as an alternative to the independent T-Test when the assumptions for the T-Test are not appropriate.

### Assumptions

- Two independent groups.
- Observations are independent.
- The dependent variable is at least ordinal.
- If you want to interpret the test as a difference in distributions/location, the groups should have reasonably similar shapes.

### When to use

**Use it when:**

- Comparing two independent groups.
- Data is ordinal or continuous but not suitable for a T-Test.
- Data contains strong skewness or outliers.
- You want a non-parametric alternative.

### When not to use

**Don't use it when:**

- Groups are paired/dependent → use Wilcoxon Signed-Rank Test.
- You have more than two independent groups → consider Kruskal-Wallis Test.
- Your data is categorical counts → consider Chi-Square.

### Business Example

An e-commerce company compares customer satisfaction ratings between:

Website A

Website B

The ratings are from 1 to 5, so they are ordinal.

### Python Example

```python
from scipy.stats import mannwhitneyu

website_A = [4, 5, 3, 4, 5, 4, 3]
website_B = [2, 3, 2, 4, 3, 2, 3]

u_stat, p_value = mannwhitneyu(
    website_A,
    website_B,
    alternative="two-sided"
)

print("U-statistic:", u_stat)
print("P-value:", p_value)
```

### Interpretation

```python
if p_value < 0.05:
    print("Significant difference between groups")
else:
    print("No significant difference")
```

A small p-value indicates evidence that the two groups differ in their distributions/ranks.

## Shapiro-Wilk Test

### Definition

The Shapiro-Wilk Test tests whether a sample is consistent with a normal distribution.

This is useful because several statistical tests, such as T-Tests and ANOVA, commonly assume approximate normality.

### Assumptions

- Observations are independent.
- Data is numeric.
- The test is intended for assessing normality of the sample.

### When to use

**Use it when:**

- You need to check normality.
- You are deciding whether a parametric test's normality assumption is reasonable.
- You have a relatively small or moderate sample.

### When not to use

**Be careful when:**

- The sample is extremely large: even tiny, practically unimportant deviations from normality can produce a very small p-value.
- You rely only on the test without examining the data visually.

It is useful to combine the test with a Q-Q plot and knowledge of the subject matter.

### Business Example

A company has employee salary data and wants to determine whether the data is approximately normally distributed before applying a statistical method that assumes normality.

### Python Example

```python
from scipy.stats import shapiro

salary = [
    45000, 47000, 48000, 50000,
    51000, 52000, 53000, 55000
]

stat, p_value = shapiro(salary)

print("Shapiro-Wilk statistic:", stat)
print("P-value:", p_value)
```

### Interpretation

```python
if p_value < 0.05:
    print("Data may not be normally distributed")
else:
    print("No strong evidence against normality")
```

Remember: p ≥ 0.05 does not prove that the data is perfectly normal. It means the test did not find strong evidence against normality.

| Test               | Main Purpose                                      | Typical Data            | Groups     |
| ------------------ | ------------------------------------------------- | ----------------------- | ---------- |
| **T-Test**         | Compare means                                     | Numerical               | 1 or 2     |
| **Z-Test**         | Test mean/proportion using normal approximation   | Numerical / proportions | 1 or 2     |
| **Chi-Square**     | Test categorical counts/association               | Categorical             | Categories |
| **ANOVA**          | Compare means                                     | Numerical               | 3+         |
| **Mann-Whitney U** | Compare two independent groups non-parametrically | Ordinal/numerical       | 2          |
| **Shapiro-Wilk**   | Test normality                                    | Numerical               | 1          |

**Most important idea**

The tests answer different questions:

- `T-Test` → Are these two means different?
- `Z-Test` → Is this mean/proportion different under a normal/Z approximation?
- `Chi-Square` → Are these categorical variables associated / do observed counts differ from expected counts?
- `ANOVA` → Are the means of 3+ groups different?
- `Mann-Whitney U` → Do two independent groups differ when a non-parametric approach is appropriate?
- `Shapiro-Wilk` → Is the data consistent with a normal distribution?
