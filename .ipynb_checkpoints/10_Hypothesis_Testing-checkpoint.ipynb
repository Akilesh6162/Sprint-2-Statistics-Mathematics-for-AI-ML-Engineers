# Notebook 10 – Hypothesis Testing

## What is Hypothesis Testing?

`Hypothesis testing` is a statistical method used to make decisions or draw conclusions about a population based on sample data.

It helps us determine whether there is enough statistical evidence to support a particular claim.

**Example**

A company claims that the average delivery time is 30 minutes.

We collect a sample of deliveries and calculate the average delivery time.

Hypothesis testing helps us determine whether the sample provides enough evidence to conclude that the actual average delivery time is different from 30 minutes.


## Null Hypothesis

The `Null Hypothesis (H₀)` is the default assumption that there is no significant effect, difference, or relationship.

It represents the claim that we test against the sample data.

**Example**

A company claims that the average delivery time is 30 minutes.

H₀: μ = 30

Where:

- H₀ → Null Hypothesis
- μ → Population mean
- 30 → Claimed average

*The Null Hypothesis (H₀) is the default assumption that there is no significant difference or effect.*

```python
# Null Hypothesis 
H0 = "Average delivery time = 30 minutes" 
print(H0)
```

## Alternative Hypothesis

The `Alternative Hypothesis` (H₁ or Hₐ) is the claim that there is a significant effect, difference, or relationship.

It is the opposite of the null hypothesis.

**Example**

If the company claims that the average delivery time is 30 minutes:

H₀: μ = 30

H₁: μ ≠ 30

This means the actual average delivery time is different from 30 minutes.

Types of Alternative Hypothesis
- H₁: μ ≠ 30    → Two-tailed
- H₁: μ > 30    → Right-tailed
- H₁: μ < 30    → Left-tailed

*The Alternative Hypothesis (H₁) is the claim that there is a significant difference or effect.*

```python
# Alternative Hypothesis
H1 = "Average delivery time != 30 minutes" 
print(H1)
```

## p-value

The `p-value` measures how much evidence the sample data provides against the null hypothesis.

A small p-value means the observed result would be relatively unlikely if the null hypothesis were true.

**Common Decision Rule**

If p-value < α
    → Reject H₀

If p-value ≥ α
    → Fail to reject H₀

Where α is the significance level.

**Example**

Suppose:

p-value = 0.03
α = 0.05

Since:

0.03 < 0.05

We reject the null hypothesis.

**Important**

A p-value does not tell us the probability that the null hypothesis is true.

*The p-value tells us how strongly the sample data conflicts with the null hypothesis.*

```python
p_value = 0.03 
alpha = 0.05 
if p_value < alpha: 
    print("Reject the Null Hypothesis") 
else: 
    print("Fail to reject the Null Hypothesis")
```

## Confidence Interval

A `confidence interval (CI)` is a range of values used to estimate an unknown population parameter.

It gives us a range in which the true population parameter is expected to lie, according to a chosen confidence level.

**Example**

Suppose the estimated average salary is:

₹50,000

and the 95% confidence interval is:

₹47,000 to ₹53,000

The interval provides a range of plausible values for the population average.

**Common Confidence Levels**

90%

95%

99%

A higher confidence level generally produces a wider interval.

*A confidence interval gives a range of values that can be used to estimate a population parameter.*

```python
import numpy as np 
from scipy import stats 
data = [45, 48, 50, 52, 55, 49, 51, 53, 47, 50] 
mean = np.mean(data) 
confidence_interval = stats.t.interval( confidence=0.95, df=len(data) - 1, loc=mean, scale=stats.sem(data) ) 
print("Mean:", mean) 
print("95% Confidence Interval:", confidence_interval)
```

## Significance Level

The `significance level (α)` is the threshold used to decide whether the evidence against the null hypothesis is statistically significant.

Common significance levels are:

α = 0.10  → 10%

α = 0.05  → 5%

α = 0.01  → 1%

The most commonly used value is:

α = 0.05

**Decision**

p-value < α
→ Reject H₀

p-value ≥ α
→ Fail to reject H₀

The significance level represents the probability of making a Type I error under the decision rule.

*The significance level (α) is the threshold used to decide whether the result is statistically significant.*

```python
alpha = 0.05 
print("Significance Level:", alpha) 
print("Significance Level:", alpha * 100, "%")
```

## Type I Error

A `Type I Error` occurs when we reject the null hypothesis even though it is actually true.

It is also called a `false positive`.

**Example**

Suppose:

H₀: A new medicine has no effect.

If we reject H₀ and conclude that the medicine has an effect when it actually does not, we have made a `Type I Error`.

**Probability of Type I Error**

The probability of a Type I Error is represented by:

**α**

For example, if:

α = 0.05

the test's Type I error rate is 5%.

```python
actual = "Machine is working correctly" 
decision = "Machine is faulty" 
if actual != decision: 
    print("Possible Type I Error: False Positive")
```

## Type II Error

A `Type II Error` occurs when we fail to reject the null hypothesis even though it is actually false.

It is also called a false negative.

**Example**

Suppose:

H₀: A new medicine has no effect.

If the medicine actually has an effect, but our test fails to detect it, we have made a Type II Error.

The probability of a Type II Error is represented by:

**β**

**Statistical Power**

The probability of correctly detecting an effect is called statistical power.

Power = 1 - β

```python
actual = "Medicine has an effect" 
decision = "No significant effect" 
if actual != decision: 
    print("Possible Type II Error: False Negative")
```

## One-Tailed Test

A `one-tailed` test checks whether a value is significantly greater than or less than a particular value.

There are two types:

Right-tailed test → >

Left-tailed test → <

**Right-Tailed Example**

Suppose a company claims that the average production is 100 units.

We want to check whether production is greater than 100.

H₀: μ ≤ 100
H₁: μ > 100

```python
sample_mean = 105 
claimed_mean = 100 
if sample_mean > claimed_mean: 
    print("Sample mean is greater than the claimed mean") 
else: 
    print("Sample mean is not greater than the claimed mean")
```

**Left-Tailed Example**

Suppose we want to check whether the average production time is less than 50 minutes.

H₀: μ ≥ 50 

H₁: μ < 50



```python
sample_mean = 45 
claimed_mean = 50 
if sample_mean < claimed_mean: 
    print("Sample mean is less than the claimed mean") 
else: 
    print("Sample mean is not less than the claimed mean")
```

## Two-Tailed Test

A two-tailed test checks whether a value is significantly different from a particular value.

It checks both directions:

Greater than

Less than

**Example**

A company claims that the average battery life is 10 hours.

We want to check whether the actual average is different from 10 hours.

H₀: μ = 10

H₁: μ ≠ 10

```python
sample_mean = 12 
claimed_mean = 10 
if sample_mean != claimed_mean: 
    print("Sample mean is different from the claimed mean") 
else: 
    print("Sample mean is equal to the claimed mean")
```

## Simple Hypothesis Testing Example Using SciPy

Now let's combine the concepts into one practical example.

**Problem**

A company claims that the average delivery time is 30 minutes.

**We collect a sample:**

data = [28, 32, 31, 29, 35, 30, 27, 33, 31, 29]

We want to test whether the average delivery time is different from 30 minutes.

```python
#Step 1: Import Libraries

import numpy as np
from scipy import stats

#Step 2: Create Data

data = [28, 32, 31, 29, 35, 30, 27, 33, 31, 29]

print("Sample Mean:", np.mean(data))

#Step 3: Perform One-Sample t-Test

t_stat, p_value = stats.ttest_1samp(data, 30)

print("t-statistic:", t_stat)
print("p-value:", p_value)

#Step 4: Make a Decision

alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to reject the Null Hypothesis")
```

#### Summary



| Topic                  | Simple Meaning             | Python Concept          |
| ---------------------- | -------------------------- | ----------------------- |
| Null Hypothesis        | Default assumption         | `H0`                    |
| Alternative Hypothesis | Opposite/alternative claim | `H1`                    |
| p-value                | Evidence against H₀        | `p_value`               |
| Confidence Interval    | Range of plausible values  | `stats.t.interval()`    |
| Significance Level     | Decision threshold         | `alpha = 0.05`          |
| Type I Error           | False positive             | Reject true H₀          |
| Type II Error          | False negative             | Fail to reject false H₀ |
| One-Tailed Test        | Tests one direction        | `>` or `<`              |
| Two-Tailed Test        | Tests both directions      | `!=`                    |
| Statistical Power      | Detecting a real effect    | `1 - beta`              |



1. `Null Hypothesis`

- **Why we need it:** It gives us a starting assumption that there is no significant difference or effect.
- **Where it is used:** Used in statistical tests to compare sample evidence against a default claim.

2. `Alternative Hypothesis`

- **Why we need it:** It represents the possibility that a real difference or effect exists.
- **Where it is used:** Used when testing whether a claim is different, greater, or smaller than the expected value.

3. `p-value`
- **Why we need it:** It helps us decide whether the sample provides enough evidence against the null hypothesis.
- **Where it is used:** Used in hypothesis tests to determine statistical significance.

4. `Confidence Interval`
- **Why we need it:** It gives a range of reasonable values instead of relying only on a single estimate.
- **Where it is used:** Used to estimate population means, proportions, and other parameters from sample data.

5. `Significance Level`
- **Why we need it:** It sets the threshold for deciding when evidence is strong enough to reject the null hypothesis.
- **Where it is used:** Commonly used in hypothesis tests, such as t-tests and chi-square tests.

6. `Type I Error`
- **Why we need it:** Understanding it helps us control the risk of incorrectly detecting an effect that does not actually exist.
- **Where it is used:** Important in medical testing, quality control, scientific research, and experiments.

7. `Type II Error`
- **Why we need it:** Understanding it helps us reduce the risk of missing a real effect.
- **Where it is used:** Important when designing experiments, medical studies, quality testing, and statistical research.

8. `One-Tailed Test`
- **Why we need it:** It is useful when we are interested in a specific direction, such as greater than or less than.
- **Where it is used:** Used when the research question has a directional expectation.

9. `Two-Tailed Test`
- **Why we need it:** It allows us to detect a difference in either direction.
- **Where it is used:** Used when we only want to know whether a value is different, without specifying whether it should be higher or lower.

**Simple overall idea**

Hypothesis testing is used when we have sample data and want to make a decision about a larger population.

**For example:**

**Company claims:** Average delivery time = 30 minutes

**We collect:** Sample delivery data

**Hypothesis testing:** Determines whether the sample gives enough evidence that the claim may be different.
