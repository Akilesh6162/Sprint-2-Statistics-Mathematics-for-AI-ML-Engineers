## Introduction to Probability

### What is Probability?

`Probability` is a number that tells us how likely something is to happen.

Probability is always between 0 and 1.

**Formula**

For equally likely outcomes:

P(A)=Number of favorable outcomes / Total number of outcomes


**Example – Coin**

A coin has two possible outcomes:

- Heads
- Tails

Probability of getting Heads:

$$ P(Heads)=\frac{1}{2}=0.5=50\% $$



```python
favorable = 1
total = 2

probability = favorable / total

print("Probability:", probability)
```

## Sample Space

### What is Sample Space?

The `sample space` is the collection of all possible outcomes of an experiment.

We usually represent sample space using:

S

**Example – Dice**

When we roll a six-sided die:


```python
S = {1, 2, 3, 4, 5, 6}

sample_space = {1, 2, 3, 4, 5, 6}

print(sample_space)
```

## Events

An `event` is one or more outcomes from the sample space that we are interested in.

**For example:**

Sample Space:
S = {1, 2, 3, 4, 5, 6}

Suppose we want to find the probability of getting an even number.

The event is:

A = {2, 4, 6}

There are 3 favorable outcomes.

Therefore:

$$ P(A)=\frac{3}{6}=0.5 $$


```python
sample_space = {1, 2, 3, 4, 5, 6}
event = {2, 4, 6}

probability = len(event) / len(sample_space)

print("Probability:", probability)
```

## Independent Events

Two events are independent when one event does not affect the probability of the other event.

**Example**

Tossing a coin twice.

Formula

For independent events:

$$ P(A\cap B)=P(A)\times P(B) $$

Here:

(A) = first event

(B) = second event

$$ (A\cap B) = A and B happen $$

Example

Probability of getting Heads twice:

$$ P(H_1\cap H_2) = \frac12\times\frac12 = \frac14 $$

So:

Probability = 0.25
            = 25%

```python
p_heads = 1 / 2

probability = p_heads * p_heads

print(probability)
```

## Dependent Events

Two events are dependent when the outcome of one event affects the probability of the other.

**Example – Drawing Cards**

Suppose we have a deck of 52 cards.

We draw one card and do not put it back.

Initially:

52 cards

Probability of getting an Ace:

$$ P(A)=\frac{4}{52} $$

After drawing an Ace, there are:

51 cards remaining

3 Aces remaining

Probability of getting another Ace:

$$ P(B|A)=\frac{3}{51} $$

The second probability changed because the first card was not replaced.

Therefore, the events are dependent.

Formula

For dependent events:

$$ P(A\cap B)=P(A)\times P(B|A) $$

## Conditional Probability

`Conditional probability` means:

The probability of an event happening given that another event has already happened.

We write:

$$ P(A|B) $$

Read this as:

Probability of A given B.

**Formula**

$$ P(A|B)=\frac{P(A\cap B)}{P(B)} $$

**Simple Example**

Suppose a class has 100 students.

- 60 students are boys
- 30 boys play cricket

What is the probability that a student plays cricket given that the student is a boy?

We only consider the boys.

Total boys = 60

Boys who play cricket = 30

Therefore:

$$ P(Cricket|Boy)=\frac{30}{60} $$ $$ =0.5 $$

So:

50%



**Important idea**

`Normal probability`:

Probability of cricket among everyone

`Conditional probability`:

Probability of cricket among boys only

## Bayes' Theorem

Bayes' Theorem helps us update a probability when we receive new information.

The formula is:

$$ P(A|B)= \frac{P(B|A)P(A)} {P(B)} $$

Where:

(P(A)) = Prior probability

(P(B|A)) = Probability of evidence given A

(P(B)) = Overall probability of evidence

(P(A|B)) = Updated probability

Probability Rules

There are several important probability rules.

Rule 1 – Probability Range

Probability must be between 0 and 1:

$$ 0\leq P(A)\leq1 $$
Rule 2 – Complement Rule

The probability that an event does not happen is:

$$ P(A^c)=1-P(A) $$


## Random Variables

A `random variable` is a variable whose value depends on the outcome of a random experiment.

For example, toss a coin.

We could define:

X = number of Heads

If we toss the coin twice:

HH → X = 2

HT → X = 1

TH → X = 1

TT → X = 0

So X can have:

X = 0, 1, 2

### Types of Random Variables

There are two main types:

**Discrete Random Variable**

Has countable values.

##### Examples:

Number of students

Number of cars

Number of heads

Number of defective products

**Continuous Random Variable**

Can take values within a range.

##### Examples:

Height

Weight

Temperature

Time

## Expected Value

`Expected value` tells us the long-run average outcome we would expect if an experiment were repeated many times.

It does not necessarily mean the outcome we will actually get in one experiment.

**Formula:**

E(X)=∑xP(x)

**Example:**

Fair Dice: (1+2+3+4+5+6)/6 = 3.5

Expected Value : 3.5

```python
outcomes = [1,2,3,4,5,6]

expected = sum(outcomes)/len(outcomes)

print(expected)
```
