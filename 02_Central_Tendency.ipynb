# Notebook 2 – Measures of Central Tendency



## Central Tendency

`Central tendency` is a way of finding a central or typical value in a dataset.

For example, suppose the marks are:

50, 60, 70, 80, 90

Instead of working with all five values, we may want to describe the dataset using one representative value.

The most common measures are:

- Mean → Average
- Median → Middle value
- Mode → Most frequently occurring value
- Weighted Mean → Average where some values have more importance
- Geometric Mean → Useful for growth rates and ratios
- Harmonic Mean → Useful for rates and ratios, especially when averaging rates

## Mean

The mean is the sum of all values divided by the number of values.

It is what we commonly call the average.

The average is:

(10 + 20 + 30) / 3 = 20

So the mean is 20.

Formula:





Example: 

salary= [40000,50000,60000,20000,80000]

Mean = (40000+50000+60000+20000+80000)/5

Mean = 50000.0

Where:

- xᵢ = individual values
- N = number of values
- x̄ = mean

```python
import numpy as np

salary = np.array([40000,50000,60000,20000,80000])

print("Average of Salary",np.mean(salary))
```

**Business Example**

Suppose a company wants to know the average amount spent by customers.

Customer spending:

- ₹500
- ₹700
- ₹600
- ₹800
- ₹400

**AI/ML Example**

Suppose a dataset contains the ages of customers:

```python
ages = [20, 25, 30, 35, 40]

average_age = np.mean(ages)

print("Average age:", average_age)
```

The mean can help during Exploratory Data Analysis (EDA).

**Interpretation**

The mean tells us the average value of the dataset.

However, mean can be strongly affected by outliers.

Example:

20, 22, 25, 27, 100

The value 100 is much larger than the other values and pulls the mean upward.

So:

*Mean works well when the data does not contain extreme outliers.*

## Median

The median is the middle value when the data is arranged in ascending or descending order.

For example:

10, 20, 30, 40, 50

The middle value is:

30

Therefore:

Median = 30

**Formula**

There are two cases.

**Odd number of observations**

If there are n values:                 `Median=middle value`

**Even number of observations**

If there are an even number of values:

                                        `Median= (middle value 1 + middle value 2)/2`
	​

**Numerical Example – Odd**

Consider:

10, 20, 30, 40, 50

There are 5 values.

The middle value is:

30

Therefore:

Median = 30

**Even Example**

Consider:

10, 20, 30, 40

The two middle values are:

20 and 30

Therefore:

Median= (20+30)/2 = 25

Mean = 25

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

median = np.median(data)

print("Median:", median)
```
